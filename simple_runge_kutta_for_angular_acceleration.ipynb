```python
import numpy as np
import matplotlib.pyplot as plt

#equation looks like:
# dw/dt = (2/M*R*R)*(tau_motor - b*w - c*w*w)

#setting arbitrary constants to investigate the nature of the w from the equation
M = 2.0
R = 0.25
constant = 2/(M*R*R)
tau_motor = 1.5
b= 0.001
c= 0.001
h = 0.01

def dw_wrt_t(w):
    return constant*(tau_motor - b*w - c*w*w)

def rungekutta(w, h): #runge kutta
    k1 = dw_wrt_t(w)
    k2 = dw_wrt_t(w + 0.5 * h * k1)
    k3 = dw_wrt_t(w + 0.5 * h * k2)
    k4 = dw_wrt_t(w + h * k3)
    return w + (h / 6.0) * (k1 + 2*k2 + 2*k3 + k4)

t = np.arange(0, 6.0, h) #time array between 0 and 6.0 in the step of h increments

w = np.zeros(len(t))
alpha = np.zeros(len(t))

for i in range(len(t)-1):
  alpha[i] = dw_wrt_t(w[i])
  w[i+1] = rungekutta(w[i], h)

plt.plot(t, w)
plt.show()

plt.plot(t, alpha)
plt.show()

```
